```python
SERVER = 'localhost'
DATABASE = 'E-Commerce'
DRIVER = 'ODBC Driver 17 for SQL Server'
connection_string = f'mssql+pyodbc://@{SERVER}/{DATABASE}?trusted_connection=yes&driver={DRIVER}'
engine = create_engine(connection_string)
```

```python
def to_date_key(dt_series):
    return pd.to_datetime(dt_series).dt.strftime('%Y%m%d').fillna(-1).astype(int)
```

```python
date_cols = [
    'order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date',
    'order_delivered_customer_date', 'order_estimated_delivery_date', 
    'review_creation_date', 'review_answer_timestamp', 'shipping_limit_date'
]

all_dates = pd.concat([df[col] for col in date_cols]).dropna().unique()
date_range = pd.date_range(start=pd.to_datetime(min(all_dates)), end=pd.to_datetime(max(all_dates)))

dim_date = pd.DataFrame({
    'DateKey': [int(d.strftime('%Y%m%d')) for d in date_range],
    'FullDate': date_range,
    'DayNumberOfWeek': date_range.dayofweek + 1,
    'DayNameOfWeek': date_range.day_name(),
    'DayNumberOfMonth': date_range.day,
    'MonthNumberOfYear': date_range.month,
    'MonthName': date_range.month_name(),
    'Quarter': date_range.quarter,
    'Year': date_range.year,
    'IsWeekend': (date_range.dayofweek >= 5).astype(int)
})

# Add missing date placeholder (-1)
missing_date = pd.DataFrame([[-1, pd.to_datetime('1900-01-01'), 0, 'N/A', 0, 0, 'N/A', 0, 0, 0]], columns=dim_date.columns)
dim_date = pd.concat([missing_date, dim_date]).drop_duplicates(subset=['DateKey'])
dim_date.to_sql('dim_date', engine, if_exists='append', index=False)
```

```python
dim_date
```

```python
customers = df[['customer_id', 'customer_unique_id', 'customer_zip_code_prefix', 'customer_city', 'customer_state']].drop_duplicates(subset=['customer_id'])
customers.to_sql('dim_customers', engine, if_exists='append', index=False)
```

```python
products = df[['product_id', 'product_category_name', 'product_category_name_english', 
               'product_name_lenght', 'product_description_lenght', 'product_photos_qty',
               'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']].drop_duplicates(subset=['product_id'])
products.to_sql('dim_products', engine, if_exists='append', index=False)
```

```python
sellers = df[['seller_id', 'seller_zip_code_prefix', 'seller_city', 'seller_state']].drop_duplicates(subset=['seller_id'])
sellers.to_sql('dim_sellers', engine, if_exists='append', index=False)
```

```python
reviews = df[['review_id', 'review_score', 'review_comment_title', 'review_comment_message', 
              'review_creation_date', 'review_answer_timestamp']].drop_duplicates(subset=['review_id'])
reviews['review_creation_date_key'] = to_date_key(reviews['review_creation_date'])
reviews['review_answer_timestamp'] = pd.to_datetime(reviews['review_answer_timestamp'])
reviews.drop(columns=['review_creation_date']).to_sql('dim_reviews', engine, if_exists='append', index=False)
```

```python
dim_cust_lookup = pd.read_sql('SELECT customer_id, CustomerKey FROM dim_customers', engine)
dim_prod_lookup = pd.read_sql('SELECT product_id, ProductKey FROM dim_products', engine)
dim_sell_lookup = pd.read_sql('SELECT seller_id, SellerKey FROM dim_sellers', engine)
dim_rev_lookup  = pd.read_sql('SELECT review_id, ReviewKey FROM dim_reviews', engine)
```

```python
fact = df.copy()
fact = fact.merge(dim_cust_lookup, on='customer_id', how='left')
fact = fact.merge(dim_prod_lookup, on='product_id', how='left')
fact = fact.merge(dim_sell_lookup, on='seller_id', how='left')
fact = fact.merge(dim_rev_lookup, on='review_id', how='left')
```

```python
fact['OrderPurchaseDateKey']    = to_date_key(fact['order_purchase_timestamp'])
fact['OrderApprovedDateKey']    = to_date_key(fact['order_approved_at'])
fact['DeliveredCarrierDateKey'] = to_date_key(fact['order_delivered_carrier_date'])
fact['DeliveredCustomerDateKey'] = to_date_key(fact['order_delivered_customer_date'])
fact['EstimatedDeliveryDateKey'] = to_date_key(fact['order_estimated_delivery_date'])
fact['ReviewCreationDateKey']   = to_date_key(fact['review_creation_date'])
fact['ReviewAnswerDateKey']     = to_date_key(fact['review_answer_timestamp'])
fact['ShippingLimitDateKey']    = to_date_key(fact['shipping_limit_date'])

# Final selection for the Fact Table
fact_orders = fact[[
    'order_id', 'order_item_id', 'order_status', 'CustomerKey', 'ProductKey', 'SellerKey', 'ReviewKey',
    'price', 'freight_value', 'payment_sequential', 'payment_type', 'payment_installments', 'payment_value',
    'OrderPurchaseDateKey', 'OrderApprovedDateKey', 'DeliveredCarrierDateKey',
    'DeliveredCustomerDateKey', 'EstimatedDeliveryDateKey', 'ReviewCreationDateKey',
    'ReviewAnswerDateKey', 'ShippingLimitDateKey'
]]

fact_orders.to_sql('fact_orders', engine, if_exists='append', index=False, chunksize=1000)
print("ETL Completed Successfully!")
```
